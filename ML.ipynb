```python
from sklearn.linear_model import LinearRegression

# Fake data: house size (sqft) -> price (in $1000s)
size = np.array([500, 750, 1000, 1250, 1500, 1750, 2000]).reshape(-1, 1)
price = np.array([150, 200, 250, 300, 340, 400, 450])

model = LinearRegression()
model.fit(size, price)                 # learn from labeled data

new_house = [[1600]]
pred = model.predict(new_house)
print(f"Predicted price for 1600 sqft: ${pred[0]:.0f}k")
```

```python
from sklearn.cluster import KMeans

# Fake customer data: (avg spend, visits/month) -- NO labels given
X = np.array([[10,1],[12,2],[11,1],[80,20],[85,22],[78,19],[40,10],[42,11]])

model = KMeans(n_clusters=3, n_init=10, random_state=0)
model.fit(X)
print("Cluster assigned to each customer:", model.labels_)
```
